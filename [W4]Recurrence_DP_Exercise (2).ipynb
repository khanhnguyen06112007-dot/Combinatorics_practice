### Question 1:
Write a pure recursive function `factorial(n)` that returns $n!$.


```python
def factorial(n):
    res = 1
    for i in range(1, n + 1):
        res *= i
    return res
```

```python
# Test cases
assert factorial(0) == 1
assert factorial(5) == 120
print('Question 1 passed!')
```

```python
def fib_rec(n):
    if n <= 1:
        return n
    return fib_rec(n - 1) + fib_rec(n - 2)
    pass
```

Find the number of binary strings of length $n$ that do not contain consecutive '1's.


```python
def no_consec_ones(n):
    if n == 0:
        return 1
    if n == 1:
        return 2
    return no_consec_ones(n - 1) + no_consec_ones(n - 2)
    pass
```

### Question 4: 
A derangement is a permutation of elements where none appear in their original position.


```python
from math import comb, factorial

def derangements(n):
    return sum(
        (-1) ** k * comb(n, k) * factorial(n - k)
        for k in range(n + 1)
    )
```

```python
# Test cases
assert derangements(3) == 2
assert derangements(4) == 9
assert derangements(6) == 265
print('Question 4 passed!')
```

### Question 5: 
The Stirling number of the second kind $S(n, k)$ is the number of ways to partition a set of $n$ objects into $k$ non-empty subsets.


```python
def stirling2(n, k):
    if n == 0 and k == 0:
        return 1
    if k == 0 or k > n:
        return 0
    return k * stirling2(n - 1, k) + stirling2(n - 1, k - 1)
```

```python
# Test cases
assert stirling2(4, 2) == 7
assert stirling2(5, 3) == 25
print('Question 5 passed!')
```

### Question 6: Fibonacci Sequence 
Pure recursion for Fibonacci is $O(2^n)$. Optimize the recurrence $f(n) = f(n-1) + f(n-2)$ using a dictionary to store previously computed values. Implement `fib_memo(n, memo)`.


```python
def fib_memo(n, memo=None):
    if memo is None:
        memo = {}

    if n in memo:
        return memo[n]

    if n <= 1:
        return n

    memo[n] = fib_memo(n - 1, memo) + fib_memo(n - 2, memo)
    return memo[n]
```

```python
# Test cases
assert fib_memo(6) == 8
assert fib_memo(50) == 12586269025 # This would hang forever in pure recursion
print('Question 6 passed!')
```

### Question 7: Tiling a $2 \times n$ Board
Find the number of ways to tile a $2 \times n$ rectangle with $2 \times 1$ dominoes. Implement `tiling_dp(n)` using a 1D DP array (or variables) to solve this iteratively.


```python
def tiling_dp(n):
    if n < 0:
        return 0
    if n <= 1:
        return 1

    prev2, prev1 = 1, 1  # số cách cho độ rộng 0 và 1
    for _ in range(2, n + 1):
        current = prev1 + prev2
        prev2, prev1 = prev1, current

    return prev1
```

```python
# Test cases
assert tiling_dp(2) == 2
assert tiling_dp(8) == 34
assert tiling_dp(40) == 165580141
print('Question 7 passed!')
```

### Question 8: Catalan Numbers 
The $n$-th Catalan number $C_n$ gives the number of valid parenthesizations of length $2n$.


```python
def catalan_dp(n):
    if n < 0:
        return 0

    catalan = [0] * (n + 1)
    catalan[0] = 1

    for size in range(1, n + 1):
        for i in range(size):
            catalan[size] += catalan[i] * catalan[size - 1 - i]

    return catalan[n]
```

```python
# Test cases
assert catalan_dp(3) == 5
assert catalan_dp(5) == 42
assert catalan_dp(12) == 208012
print('Question 8 passed!')
```

### Question 9: Longest Common Subsequence 
Given strings `text1` and `text2`, find the length of their longest common subsequence.


```python
def lcs(text1, text2):
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i - 1] == text2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    return dp[m][n]
```

```python
# Test cases
assert lcs('abcde', 'ace') == 3
assert lcs('abc', 'def') == 0
print('Question 9 passed!')
```

### Question 10: Matrix Chain Multiplication 
Given an array `p` representing matrix dimensions, find the minimum scalar multiplications needed to multiply them. The recurrence splits the sequence at $k$: $M[i,j] = \min_{i \le k < j} (M[i,k] + M[k+1,j] + p_{i-1}p_k p_j)$. Implement `matrix_chain(p)` using DP.


```python
def matrix_chain(p):
    n = len(p) - 1  
    if n <= 1:
        return 0

    dp = [[0] * n for _ in range(n)]
    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            dp[i][j] = min(
                dp[i][k] + dp[k + 1][j] + p[i] * p[k + 1] * p[j + 1]
                for k in range(i, j)
            )

    return dp[0][n - 1]
```

```python
# Test cases
assert matrix_chain([40, 20, 30, 10, 30]) == 26000 # the matrices are 40x20, 20x30, 30x10, 10x30
print('Question 10 passed!')
```
